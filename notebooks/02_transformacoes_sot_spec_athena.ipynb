# Notebook 02 — Transformações SOT/Silver e SPEC/Gold

**Ambiente:** Amazon Athena; execute uma célula SQL por vez.  
**Entrada:** tabelas `pesquisas_<ano>` do Glue Catalog.  
**Saída:** tabelas SOT e SPEC em Parquet no S3.

SOT padroniza nomes técnicos; SPEC atribui nomes de negócio. Antes de repetir um CTAS, remova a tabela e limpe somente a pasta S3 daquela camada e ano.

## 2023 — SOT / Silver
